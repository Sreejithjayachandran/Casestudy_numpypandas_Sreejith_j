```python
#read csv file
import pandas as pd
print(pd.__version__)
filepath='https://raw.githubusercontent.com/laxmimerit/All-CSV-ML-Data-Files-Download/master/IMDB-Movie-Data.csv'
df=pd.read_csv(filepath)
df
```

```python
#check duplicate
df.duplicated()
```

```python
# Removing the duplicates
df.drop_duplicates()
```

```python
#Renaming the column

df=df.rename(columns={"Title": "Movie_Title"})
```

```python
# finding unique genres
unique = df["Genre"].unique()
print(unique)
```

```python
# Extract number of genres per movie

df["Genre_Count"] = df["Genre"].str.split("|").str.len()
print(df[["Movie_Title", "Genre", "Genre_Count"]].head())
```

```python
# Handling missing values in Revenue(millions) and metascore

df['Revenue (Millions)'] = df['Revenue (Millions)'].fillna(
    df['Revenue (Millions)'].median()
)


df['Metascore'] = df['Metascore'].fillna(
    df['Metascore'].median()
)

print(df[['Revenue (Millions)', 'Metascore']].isnull().sum())


```

```python
# Top 5 movies by Rating
top_movies=df.nlargest(5,'Rating')
print(top_movies[["Movie_Title", "Rating"]])
```

```python
#Movies with highest/lowest revenue

# movie with  highest revenue
highest_revenue_movie = df.loc[df["Revenue (Millions)"].idxmax()]

# movie with lowest revenue
lowest_revenue_movie = df.loc[df["Revenue (Millions)"].idxmin()]

print(highest_revenue_movie[["Movie_Title","Revenue (Millions)"]])
print("\n")
print(lowest_revenue_movie[["Movie_Title","Revenue (Millions)"]])
```

```python
# Avg. rating by Director
Avg_rating_by_director = df.groupby("Director")["Rating"].mean()
print(Avg_rating_by_director)
```

```python
# Movies per year
movies_per_year = df.groupby("Year")["Movie_Title"].count()
print(movies_per_year)
```

```python
# Movies with rating > 8 & revenue > 100M
movies_rating = df[(df["Rating"] > 8) & (df["Revenue (Millions)"] > 100)]
print(movies_rating [["Movie_Title", "Rating", "Revenue (Millions)"]])
```

```python
# Group by Director: avg Rating, Revenue
director_group = df.groupby("Director")[["Rating", "Revenue (Millions)"]].mean()
print(director_group)
```

```python
# Group by Year: count, avg Rating

year_group=df.groupby("Year")["Rating"].agg(["count","mean"])
print(year_group)
```

```python
# Create a new column Profitability

# a) Revenue divided by Duration
df["Profitability"] =df["Revenue (Millions)"] / df["Runtime (Minutes)"]
print(df[["Movie_Title", "Revenue (Millions)", "Profitability"]].head())
```

```python
# Find top 10 most profitable movies
df_profitable = df.nlargest(10, "Profitability")
print(df_profitable[["Movie_Title",  "Profitability"]])
#
```

```python
# Create a column that categorizes movies by rating:
# a)Rating < 6: Poor
# b) 6 ≤ Rating < 7.5: Average
# c) Rating ≥ 7.5: Excellent

import numpy as np

conditions = [
    df["Rating"] < 6,
    (df["Rating"] >= 6) & (df["Rating"] < 7.5),
    df["Rating"] >= 7.5,
]


choices = ["Poor", "Average", "Excellent"]


df["new_rating"] = np.select(conditions, choices, default="Unknown")

print(df[["Movie_Title", "Revenue (Millions)", "new_rating"]].head(1001))

```

```python
# Create a custom function to score movies based on:
# i) Apply row['Rating'] * 10 + row['Revenue (Millions)'] as High score
# Apply a function to dataset to assign the score

def calculate_high_score(row):

    rating = row["Rating"]
    revenue = row["Revenue (Millions)"]


    return (rating * 10) + revenue



df["High_Score"] = df.apply(calculate_high_score, axis=1)

print(df[["Movie_Title", "Rating", "Revenue (Millions)", "High_Score"]].head())

```
